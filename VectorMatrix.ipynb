Numpy 는 데이터 형을 강제로 단순화하여 연산속도를 높여주는 라이브러리 입니다.

```python
import numpy as np                      # numpy 라이브러리       <1>
import time                             # time 라이브러리

x = np.random.random(100000000)         # 1 억개의 랜덤 수

# ---- Python 의 for 문 사용 ---------------------------------
start = time.time()                     # 현재의 시간을 기록 합니다.
y = 0
for r in x:                             # 반복문 for
    y += r
z = y/len(x)

print('Python for 평균값 계산 시간:   ', time.time() - start)

# ---- Python의 sum() 사용 ----------------------------------
start = time.time()                     # 현재의 시간을 기록 합니다.
sum(x) / len(x)                         # 더하기 함수
print('Python sum(x) 평균값 계산 시간:', time.time() - start)

# ---- Numpy 의 mean() 사용 ---------------------------------
start = time.time()                     # 현재의 시간을 기록 합니다.
np.mean(x)                              # numpy 평균값
print('Numpy  평균값 계산 시간:       ', time.time() - start)
```

Python 에서의 행열을 NumPy 행열로 변환

```python
import numpy as np        # numpy 라이브러리       <2>

# Python 행열(리스트 표현)
Python_Matrix = [[1,2,3],[4,5,6],[7,8,9]]
print('Python_Matrix:\n',Python_Matrix)

# Python 행열을 Numpy 행열로 변환
Numpy_Matrix = np.array(Python_Matrix)
print('Numpy_Matrix:\n',Numpy_Matrix)

s = (3,2)                               # 행열의 형태(행, 열) 지정 (튜플)
A = np.zeros(s)                         # 모든 요소를 0 으로 채움
B = np.ones(s, dtype=np.uint16)         # 모든 요소를 1 로 채움
C = np.full(s, np.pi, dtype=np.float64) # 지정한 수(np.pi = 3.141592)로 채움
D = np.random.randn(3,3)                # 모든 요소를 랜덤 수로 채움
E = np.empty(s, dtype=np.uint64)        # 모든 요소를 채우지 않음

print('0 으로 채움\n',A)
print('1 으로 채움\n',B)
print('3.14로 채움\n',C)
print('난수로 채움\n',D)
print('빈공간으로 설정\n',E)
```

행열 곱셈 3

[M행, P열] x [P행, N열] ---> [M행, N열]

image.png

numpy의 1차원 배열은 "행벡터"도 "열벡터"도 아닌, 그냥 방향 없는 1차원 데이터이기 때문에 전치가 의미를 갖지 못한다.

```python
import numpy as np                # numpy 라이브러리       <8>

a = np.array([1, 2, 3])        # 1 차원배열이나 2 차원으로 간주한다.
b = np.array([[4],
              [5],
              [6]])

c = np.matmul(a, b)            # 스칼라 데이터가 된다.
print('c = ', c, c.shape)

d = np.matmul(b, a.reshape(1, -1)) # Reshape 'a' to (1, 3) for outer product
print('d = ', d, d.shape)
```

```python
import numpy as np                # numpy 라이브러리       <13>

# 1 학기성적  국어 영어 수학 사회
y = np.array([[[55, 92, 23, 11],    # 철수
               [10, 36, 15, 78],    # 영희
               [82, 18, 75, 13]],   # 만배

# 2 학기성적  국어 영어 수학 사회
              [[16, 14, 21, 19],    # 철수
               [12, 20, 43, 54],    # 영희
               [62, 97, 33, 0]]])   # 만배

print('1,2학기 전체 학생 과목별 평균:\n',np.mean(y,axis=0))
print('1,2학기 전체 학생 과목별 최고:\n',np.max(y,axis=1))
print('1,2학기 학생별 최고:\n',np.max(y,axis=2, keepdims=True))
print('1학기 과목별 최저:', np.min(y[0],axis=0))
print('2학기 학생별 최저:', np.min(y[1],axis=1))
print('영희의 2 학기 최저:', np.min(y[1],axis=1)[1])
print('1,2학기 전체 학생 최고점 점수:', np.max(y))
print('최저 점수의 학기, 학생, 과목:', np.where(y==0))             # 1학기, 만배, 사회
```

```python
import numpy as np                # numpy 라이브러리       <14>

# 1 학기성적  국어 영어 수학 사회
y = np.array([[[55, 92, 23, 11],    # 철수
               [10, 36, 15, 78],    # 영희
               [82, 18, 75, 13]],   # 만배

# 2 학기성적  국어 영어 수학 사회
              [[16, 14, 21, 19],    # 철수
               [12, 20, 43, 54],    # 영희
               [62, 97, 33, 0]]])   # 만배

a = np.mean(y,axis=0)               # 1 학기와 2 학기의 학생별, 과목별 평균 성적
print(a, a.shape)                   # 1 학기와 2 학기의 구분이 없어지고 학생과 과목이 남음
                                    # Axis 0(차원축)이 사라짐

b = np.mean(a,axis=0)               # 과목별 성적 평균
print(b, b.shape)                   # 학생의 구분이 없어지고 과목만 남음,
                                    # Axis 0(학생축)이 사라짐

c = np.mean(b,axis=0)               # 전체 성적 평균
print(c, c.shape)                   # 과목의 구분이 없어지고 전체(1, 2학기, 학생, 과목)
                                    # 평균만 남음, Axis 0(과목축)이 사라짐
```
