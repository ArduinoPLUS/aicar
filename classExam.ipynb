```python
# Class 예제                                                      <classExam.py>

import numpy as np

#--- 창고 건설 class -----------------------------------------------------------
class warehouse:
    # class 초기화
    def __init__(self, floor, height, width):    # 창고 건설

        # 스트링 또는 Python 객체를 저장 가능한 NumPy 배열 초기화
        self.data = np.full((floor, height, width), '', dtype=object)
        self.stock_count = 0                     # 창고 재고 개수

    # 상품 저장하기 메서드(Method)
    def save(self, L, H, W, Box):                # 저장할 위치와 상품
        self.data[L, H, W] = Box
        self.stock_count += 1                    # 상품 재고 증가

    # 상품 가져오기 메서드(Method)
    def get(self, L, H, W):                      # 가져올 위치
        self.stock_count -= 1                    # 상품 재고 감소
        return self.data[L, H, W]                # 상품 출고

#--- 창고 건설(인스턴스 생성) --------------------------------------------------
DAEGU_fruits = warehouse(3, 200, 100)    # 인스턴스 생성 - 대구 3 층, 200, 100 과일 창고 건설
SUWON_fruits = warehouse(2, 100, 90)     # 인스턴스 생성 - 수원 2 층, 100, 90 과일 창고 건설
SEOUL_computer = warehouse(20, 20, 10)   # 인스턴스 생성 - 서울 15 층, 20, 10 컴퓨터 창고 건설
BUSAN_computer = warehouse(15, 12, 11)   # 인스턴스 생성 - 부산 20 층, 12, 11 컴퓨터 창고 건설
#-------------------------------------------------------------------------------
# 과일 창고 1층, 세로 2 번, 가로 3 번에 5 저장
DAEGU_fruits.save(1,2,3, '사과')                     # save 메서드

# 컴퓨터 창고 9층, 세로 8 번, 가로 7 번에 6 저장
SEOUL_computer.save(9,8,7, 'NoteBook')               # save 메서드

# 대구 과일 창고 1 층 가로 2 번, 세로 3번 상품 가져오기
basket = DAEGU_fruits.get(1,2,3)                     # get 메서드

# 대구에서 가져온 상품 확인
print(basket)

# 대구에서 가저온 과일을 수원 과일 창고 2층, 가로 3, 세로 1에 저장
SUWON_fruits.save(1,3,1, basket)                     # save 메서드

# 서울 컴퓨터 창고 9층, 가로 8, 세로 7 에 저장된 상품 확인
print(SEOUL_computer.get(9,8,7))                     # get 메서드

# 창고의 재고 현황 출력
print(DAEGU_fruits.stock_count)                      # 대구 과일창고 재고 보유현황
print(SUWON_fruits.stock_count)                      # 수원 과일창고 재고 보유현황
print(SEOUL_computer.stock_count)                    # 서울 컴퓨터창고 재고 보유현황
print(BUSAN_computer.stock_count)                    # 부산 컴퓨터창고 재고 보유현황
```
