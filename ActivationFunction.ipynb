```python
# Sigmoid 활성화 함수                               0 <ActivationFunction.ipynb>

import matplotlib.pyplot as plt
import numpy as np

# Sigmid 함수
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

# Sigmoid 함수의 미분
def sigmoid_derivative(x):
  s = sigmoid(x)
  return s * (1 - s)

# x 값 범위 설정
x = np.linspace(-6, 6, 100)

# sigmoid 와 sigmoid 미분 값 계산
y_sigmoid = sigmoid(x)
y_sigmoid_derivative = sigmoid_derivative(x)

# 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(x, y_sigmoid, label='Sigmoid Function', color='blue')
plt.plot(x, y_sigmoid_derivative, label='Sigmoid Derivative', color='red', linestyle='--')
plt.title('Sigmoid Function and its Derivative')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
# Set the x-axis ticks to 1 unit intervals
plt.xticks(np.arange(min(x), max(x) + 1, 1))

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```

```python
# Tanh 활성화 함수                               1 <ActivationFunction.ipynb>

import numpy as np
import matplotlib.pyplot as plt

# 1. tanh 함수
def tanh(x):
    return np.tanh(x)

# 2. tanh 함수의 미분
# tanh'(x) = 1 - tanh(x)^2
def tanh_derivative(x):
    return 1 - np.tanh(x)**2

# x 값 범위 설정
x = np.linspace(-5, 5, 400) # -5부터 5까지 400개의 점 생성

# y (tanh) 및 dy (tanh 미분) 값 계산
y = tanh(x)
dy = tanh_derivative(x)

# 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='tanh Function', color='blue')
plt.plot(x, dy, label='tanh Derivative', color='red', linestyle='--')
plt.title('Hyperbolic Tangent (tanh) Function and its Derivative')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
# Set the x-axis ticks to 1 unit intervals
plt.xticks(np.arange(min(x), max(x) + 1, 1))
# Set the aspect ratio to be equal
#plt.gca().set_aspect('equal', adjustable='box')
plt.tight_layout()

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```

```python
# ELU 활성화 함수                                   2 <ActivationFunction.ipynb>

import numpy as np
import matplotlib.pyplot as plt

# ELU 함수
def elu(x, alpha=1.0):
  return np.where(x > 0, x, alpha * (np.exp(x) - 1))

# ELU 함수 미분
def elu_derivative(x, alpha=1.0):
  return np.where(x > 0, 1, alpha * np.exp(x))

# x 값 범위 설정
x = np.linspace(-4, 4, 100)

# y 값, ELU 미분 값
y_elu = elu(x)
y_elu_prime = elu_derivative(x)

# 그래프 그리기
plt.figure(figsize=(10, 5))
plt.plot(x, y_elu, label='ELU Function', color='blue')
plt.plot(x, y_elu_prime, label='ELU Derivative', color='red', linestyle='--')
plt.title('ELU Activation Function and its Derivative')
plt.xlabel('x')
plt.ylabel('Value')
plt.grid(True)
# Set the x-axis ticks to 1 unit intervals
plt.xticks(np.arange(min(x), max(x) + 1, 1))
# Set the aspect ratio to be equal
#plt.gca().set_aspect('equal', adjustable='box')
plt.tight_layout()
plt.legend()
plt.show()
```

```python
# ReLU 활성화 함수                                   3 <ActivationFunction.ipynb>

import numpy as np
import matplotlib.pyplot as plt

# 1. ReLU 함수 정의
def relu(x):
    return np.maximum(0, x)

# 2. ReLU 함수의 미분 정의
# x > 0 이면 1, x <= 0 이면 0
def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# x 값 범위 설정
x = np.linspace(-5, 5, 400) # -5부터 5까지 400개의 점 생성

# y (ReLU) 및 dy (ReLU 미분) 값 계산
y = relu(x)
dy = relu_derivative(x)

# 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='ReLU Function', color='blue')
plt.plot(x, dy, label='ReLU Derivative', color='red', linestyle='--')
plt.title('Rectified Linear Unit (ReLU) Function and its Derivative')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
# Set the x-axis ticks to 1 unit intervals
plt.xticks(np.arange(min(x), max(x) + 1, 1))
plt.tight_layout() # 그래프 간 간격 최적화
plt.show()
```

```python
# Leaky ReLU 활성화 함수                             4 <ActivationFunction.ipynb>

import numpy as np
import matplotlib.pyplot as plt

# 1. Leaky ReLU 함수 정의
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# 2. Leaky ReLU 함수의 미분 정의
# x > 0 이면 1, x <= 0 이면 alpha (기본값 0.01)
def leaky_relu_derivative(x, alpha=0.01):
    return np.where(x > 0, 1, alpha)

# x 값 범위 설정
x = np.linspace(-5, 5, 400) # -5부터 5까지 400개의 점 생성

# y (Leaky ReLU) 및 dy (Leaky ReLU 미분) 값 계산
y = leaky_relu(x)
dy = leaky_relu_derivative(x)

# 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Leaky ReLU Function (alpha=0.01)', color='blue')
plt.plot(x, dy, label='Leaky ReLU Derivative (alpha=0.01)', color='red', linestyle='--')
plt.title('Leaky Rectified Linear Unit (Leaky ReLU) Function and its Derivative')
plt.xlabel('x')
plt.ylabel('y / Derivative Value')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.tight_layout() # 그래프 간 간격 최적화
plt.show()
```

### Softmax 활성화 함수

**Softmax 함수**는 신경망의 마지막 레이어에서 주로 사용되는 활성화 함수로, 다중 클래스 분류 문제에서 각 클래스에 속할 확률을 예측하는 데 사용됩니다. 로지스틱 회귀를 다중 클래스로 확장한 것이라고 볼 수 있습니다.

**주요 특징:**

1.  **확률 분포 출력**: 입력 벡터의 각 요소를 0과 1 사이의 값으로 변환하며, 이 값들의 총합은 항상 1이 됩니다. 따라서 출력을 각 클래스에 대한 확률로 해석할 수 있습니다.
2.  **"가장 큰" 값 강조**: 입력 값 중 가장 큰 값(가장 높은 점수)에 해당하는 클래스의 확률을 더 높게 만들고, 다른 클래스의 확률은 상대적으로 낮게 만들어 분류를 명확하게 합니다.
3.  **수학적 표현**: $K$개의 클래스에 대한 입력 벡터 $\mathbf{z} = [z_1, z_2, \dots, z_K]$가 주어졌을 때, Softmax 함수는 각 클래스 $i$에 대한 확률 $P(y=i|\mathbf{z})$를 다음과 같이 계산합니다.

    $$P(y=i|\mathbf{z}) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$

    여기서 $e$는 자연 로그의 밑(약 2.718)이며, $z_i$는 $i$번째 클래스에 대한 모델의 로짓(logit) 또는 점수입니다.

**사용 예시:**

*   이미지 분류: 주어진 이미지가 '고양이', '개', '새' 중 어느 클래스에 속할 확률이 가장 높은지 예측할 때 사용됩니다.
*   자연어 처리: 단어가 특정 품사에 속할 확률이나 문장이 특정 감성(긍정, 부정, 중립)에 속할 확률을 예측할 때 사용됩니다.

```python
import numpy as np

# Softmax 함수 정의
def softmax(x):
  # 오버플로우 방지를 위해 최대값(max)을 빼줌
  e_x = np.exp(x - np.max(x))
  return e_x / e_x.sum(axis=0)

# 예시 입력 (로짓 또는 점수)
logits = np.array([3.0, 1.0, 0.2])

# Softmax 함수 적용
probabilities = softmax(logits)

print("입력 로짓:", logits)
print("Softmax 출력 확률:", probabilities)
print("확률의 합:", probabilities.sum())

# 다른 예시 (더 큰 로짓 값)
logits_2 = np.array([10.0, 2.0, 1.0])
probabilities_2 = softmax(logits_2)

print("\n다른 입력 로짓:", logits_2)
print("Softmax 출력 확률:", probabilities_2)
print("확률의 합:", probabilities_2.sum())
```

### Softmax 함수의 자코비안 행렬 (Jacobian Matrix)

Softmax 함수는 입력 벡터 $\mathbf{z} = [z_1, z_2, \dots, z_K]$를 받아서 출력 벡터 $\mathbf{s} = [s_1, s_2, \dots, s_K]$를 생성합니다. 여기서 각 $s_i$는 다음과 같습니다.

$$s_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$

Softmax 함수의 자코비안 행렬은 출력 벡터 $\mathbf{s}$를 입력 벡터 $\mathbf{z}$에 대해 미분한 것으로, $K \times K$ 행렬 형태로 표현됩니다. 각 요소 $J_{ij}$는 $s_i$를 $z_j$에 대해 편미분한 것입니다.

$$J_{ij} = \frac{\partial s_i}{\partial z_j}$$

이 미분은 두 가지 경우로 나누어 계산할 수 있습니다.

**1. $i = j$ 인 경우:**

$s_i$를 $z_i$에 대해 미분합니다. 몫의 미분법($(\frac{f}{g})' = \frac{f'g - fg'}{g^2}$)을 사용합니다.

*   $f = e^{z_i} \implies f' = e^{z_i}$
*   $g = \sum_{k=1}^{K} e^{z_k} \implies g' = e^{z_i}$ (나머지 $z_k$에 대한 미분은 0이 됩니다.)

$$ \frac{\partial s_i}{\partial z_i} = \frac{e^{z_i}(\sum_{k=1}^{K} e^{z_k}) - e^{z_i}(e^{z_i})}{(\sum_{k=1}^{K} e^{z_k})^2} $$

분모의 합을 $S$라고 하면 ($S = \sum_{k=1}^{K} e^{z_k}$):

$$ = \frac{e^{z_i}S - (e^{z_i})^2}{S^2} = \frac{e^{z_i}}{S} - \left(\frac{e^{z_i}}{S}\right)^2 = s_i - s_i^2 = s_i(1 - s_i) $$

**2. $i \neq j$ 인 경우:**

$s_i$를 $z_j$에 대해 미분합니다.

*   $f = e^{z_i} \implies f' = 0$ (여기서는 $z_j$에 대해 미분하므로 $e^{z_i}$는 상수 취급)
*   $g = \sum_{k=1}^{K} e^{z_k} \implies g' = e^{z_j}$

$$ \frac{\partial s_i}{\partial z_j} = \frac{0 \cdot (\sum_{k=1}^{K} e^{z_k}) - e^{z_i}(e^{z_j})}{(\sum_{k=1}^{K} e^{z_k})^2} $$

$$ = \frac{-e^{z_i}e^{z_j}}{S^2} = -\left(\frac{e^{z_i}}{S}\right)\left(\frac{e^{z_j}}{S}\right) = -s_i s_j $$

**결론적으로 Softmax 함수의 자코비안 행렬의 각 요소는 다음과 같습니다:**

$$ \frac{\partial s_i}{\partial z_j} = \begin{cases} s_i(1 - s_i) & \text{if } i = j \\ -s_i s_j & \text{if } i \neq j \end{cases} $$

이를 행렬 형태로 표현하면 다음과 같습니다.

$$ \mathbf{J} = \begin{pmatrix}
  s_1(1-s_1) & -s_1s_2 & \dots & -s_1s_K \\
  -s_2s_1 & s_2(1-s_2) & \dots & -s_2s_K \\
  \vdots & \vdots & \ddots & \vdots \\
  -s_Ks_1 & -s_Ks_2 & \dots & s_K(1-s_K)
\end{pmatrix} $$

또는 Kronecker 델타 함수 $\delta_{ij}$를 사용하여 다음과 같이 간결하게 표현할 수도 있습니다.

$$ \frac{\partial s_i}{\partial z_j} = s_i(\delta_{ij} - s_j) $$

여기서 $\delta_{ij}$는 $i=j$일 때 1이고, $i \neq j$일 때 0입니다.

```python
import numpy as np

def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum(axis=0)

def softmax_jacobian(x):
    s = softmax(x)
    K = len(s)
    jacobian_matrix = np.zeros((K, K))

    for i in range(K):
        for j in range(K):
            if i == j:
                jacobian_matrix[i, j] = s[i] * (1 - s[i])
            else:
                jacobian_matrix[i, j] = -s[i] * s[j]
    return jacobian_matrix

# 예시 입력
logits = np.array([3.0, 1.0, 0.2])

# Softmax 출력 계산
probabilities = softmax(logits)
print("Softmax 출력 확률:", probabilities)

# Softmax 자코비안 행렬 계산
jacobian = softmax_jacobian(logits)
print("\nSoftmax 자코비안 행렬:")
print(jacobian)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax_2class(x):
    """2-class Softmax: class0 logit=0, class1 logit=x"""
    logits = np.stack([np.zeros_like(x), x], axis=0)
    exp = np.exp(logits)
    return exp[1] / np.sum(exp, axis=0)

def softmax_derivative(x, epsilon=1e-5):
    """수치 미분"""
    return (softmax_2class(x + epsilon) - softmax_2class(x - epsilon)) / (2 * epsilon)

x = np.linspace(-5, 5, 500)

plt.figure(figsize=(12, 7))
plt.plot(x, softmax_2class(x), label='Softmax P(class=1)', linewidth=3.5, color='#1f77b4')
plt.plot(x, softmax_derivative(x), label='Derivative d(Softmax)/dx',
         linewidth=3, color='#d62728', linestyle='--')

plt.title('Softmax Function and Its Derivative\n(2-class case)', fontsize=16)
plt.xlabel('Logit value (x)', fontsize=13)
plt.ylabel('Value', fontsize=13)
plt.grid(True, alpha=0.3)
plt.axhline(0, color='black', lw=0.8)
plt.axvline(0, color='black', lw=0.8)
plt.legend(fontsize=12.5)
plt.ylim(-0.05, 1.05)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax(logits):
    exp = np.exp(logits - np.max(logits, axis=0, keepdims=True))
    return exp / np.sum(exp, axis=0, keepdims=True)

x = np.linspace(-5, 5, 500)
n_classes = 10
logits = np.zeros((n_classes, len(x)))
logits[0] = x                     # Class 0의 logit만 변화

probs = softmax(logits)

# 수치 미분
epsilon = 1e-5
logits_plus = logits.copy()
logits_plus[0] += epsilon
probs_plus = softmax(logits_plus)
derivatives = (probs_plus - probs) / epsilon

# 그래프
fig, axs = plt.subplots(2, 1, figsize=(14, 10))

# Softmax Probability
for i in range(n_classes):
    axs[0].plot(x, probs[i], label=f'Class {i}',
                linewidth=3 if i==0 else 1.8, alpha=1 if i==0 else 0.75)

axs[0].set_title('10-Class Softmax Probabilities\n(Varying logit of Class 0, others = 0)', fontsize=16)
axs[0].set_xlabel('Logit value for Class 0', fontsize=13)
axs[0].set_ylabel('Probability', fontsize=13)
axs[0].grid(True, alpha=0.3)
axs[0].legend(fontsize=11)

# Derivative
for i in range(n_classes):
    axs[1].plot(x, derivatives[i], label=f'dP{i}/d(logit0)',
                linewidth=3 if i==0 else 1.8, alpha=1 if i==0 else 0.75)

axs[1].set_title('Derivatives of Probabilities w.r.t. Class 0 logit', fontsize=16)
axs[1].set_xlabel('Logit value for Class 0', fontsize=13)
axs[1].set_ylabel('Derivative', fontsize=13)
axs[1].grid(True, alpha=0.3)
axs[1].legend(fontsize=11)

plt.tight_layout()
plt.show()
```
