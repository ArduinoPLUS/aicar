1 개의 y 리스트 데이터로 그래프를 그립니다. 정의역 데이데이터 x 가 주어지지 않으면 0 부터 정수로 자동 생성됩니다. 색상 지정이 없으면 blue 입니다.

```python
import matplotlib.pyplot as plt                                 # <1>

y = [3,2,5,-1,6,0,3]           # 7 개 데이터로 구성된 1 개의 List
plt.plot(y)                    # 0 - 6 정의역 7 개 자동 생성
plt.show()                     # 그래프의 기본 색상은 blue
```

같은 개수로 구성된 여러개의 리스트, 색상은 blue 부터 시작하며 10 개의 색상이  반복됩니다.

선의 형태와 마크를 조합할 수 있습니다.

선의 형태는 기본 연속('-'), 대시('--'), 점선(':') 이며 마커는 ('o'),('s'),('x')를 선택할 수 있습니다.

마크만 있을때 선은 그려지지 않습니다.

```python
import matplotlib.pyplot as plt                                 # <2>

a = [93,92,95,91,96,90,92]           # List 1
b = [83,82,85,81,86,80,82]           # List 2
c = [73,72,75,71,76,70,72]           # List 3
d = [63,62,65,61,66,60,62]           # List 4
e = [53,52,55,51,56,50,52]           # List 5
f = [43,42,45,41,46,40,42]           # List 6
g = [33,32,35,31,36,30,32]           # List 7
h = [23,22,25,21,26,20,22]           # List 8
i = [13,12,15,11,16,10,12]           # List 9
j = [ 3, 2, 5, 1, 6, 0, 2]           # List 10
k = [-7,-8,-5,-9,-6,-10,-8]

plt.plot(a)                          # blue
plt.plot(b,linewidth=1)              # orange
plt.plot(c,linewidth=3)              # green
plt.plot(d,linewidth=5)              # red
plt.plot(e,linewidth=10)             # puple
plt.plot(f, '-')                     # brown
plt.plot(g, '--')                    # pink
plt.plot(h, ':')                     # gray
plt.plot(i, '-.')                    # olive
plt.plot(j)                          # cyan
plt.plot(k, 'k',linewidth=5)         # black width=5
plt.plot(k, 'w',linewidth=1)         # white width=1
plt.show()
```

1문자 컬러코드

'b':blue, 'g':green, 'r':red, 'c':cyan, 'm':magenta, 'y':yellow, 'k':white

라인 스타일
'-': 연속실선, '--': 대시선, ':':점선

마커모양
'o':동그라미, 's':사각형,  '^': 삼각형, 'x':알파벳, x

```python
import matplotlib.pyplot as plt                                 # <3>

a = [55, 105,45, 75]           # List 1
b = [50, 100,40, 70]           # List 2
c = [45, 95, 35, 65]           # List 3
d = [40, 90, 30, 60]           # List 4
e = [35 ,85, 25, 55]           # List 5
f = [30, 80, 20, 50]           # List 6
g = [25, 75, 15, 45]           # List 7
h = [20, 70, 10, 40]           # List 8
i = [15, 65,  5, 35]           # List 9
j = [10, 60,  0, 30]           # List 10

plt.plot(a, 'b.-', label= 'blue point')             # blue    point     '.'
plt.plot(b, 'g1-', label= 'tri_down')               # green   tri_down
plt.plot(c, 'r2-', label= 'tri_up')                 # red     tri_up
plt.plot(d, 'c3-', label= 'tri_left')               # cyan    tri_left
plt.plot(e, 'm4-', label= 'tri_right')              # magenta tri_right
plt.plot(f, 'y+-', label= 'plus +')                 # yellow  plus      '+'
plt.plot(g, 'kx-', label= 'x')                      # black   x         'x'
plt.plot(h, 'k|-', label= 'vline |')                # black   vline     '|'
plt.plot(i, 'k_-', label= 'hline _')                # black   hline     '_'
plt.plot(j, 'r-', linewidth=5, label= 'pixel')      # red
plt.plot(j, 'w,')                                   # white pixel
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt                                 # <3>

a = [100,150,90,120,120]           # List 1
b = [90,140, 80,110,110]           # List 2
c = [80,130, 70,100,100]           # List 3
d = [70,120, 60, 90, 90]           # List 4
e = [60,110, 50, 80, 80]           # List 5
f = [50,100, 40, 70, 70]           # List 6
g = [40, 90, 30, 60, 60]           # List 7
h = [30, 80, 20, 50, 50]           # List 8
i = [20, 70, 10, 40, 40]           # List 9
j = [10, 60,  0, 30, 30]           # List 10
k = [ 0, 50,-10, 20, 20]           # List 11
l = [-10,40,-20, 10, 10]           # List 12

plt.plot(a, 'bo-')                 # blue    point     '.'
plt.plot(b, 'gv-')                 # green   tri_down
plt.plot(c, 'r^-')                 # red     triangle up
plt.plot(d, 'c<-')                 # cyan    triangle left
plt.plot(e, 'm>-')                 # magenta triangle right
plt.plot(f, 'ys-')                 # yellow  square
plt.plot(g, 'kp-', markersize=9)                                              # black pentagon
plt.plot(h, 'k*-', markersize=19, markeredgewidth=1, markerfacecolor='red')   # black star
plt.plot(i, 'kh-', markerfacecolor='orange')                                  # black hexagon 1
plt.plot(j, 'rH-', markersize=9, markerfacecolor='black')                     # red   hexagon 2
plt.plot(k, 'bD-', markersize=9, markeredgewidth=3, markerfacecolor='yellow') # blue  diamond
plt.plot(l, 'gd-', markerfacecolor='white')                                # green thin diamond

plt.show()
```

x 리스트 데이터와 y 리스트 데이터가 같이 주어진 그래프 입니다.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]   # x 좌표
y = [3,2,5,4,6]   # y 좌표

plt.plot(x, y, label = 'Trend')

plt.title(f'Graph')
plt.xlabel('x')
plt.ylabel('f(x)')

plt.axvline(x=1.3, color='orange', linestyle='--', label=f'V line')
plt.axhline(y=3.2, color='red', linestyle=':', label=f'H line')
plt.grid()
plt.legend()

plt.show()
```

```python
import numpy as np                                         # <4>
import matplotlib.pyplot as plt

x = np.arange(0, 10, 0.1)  # 0 부터 10 까지 0.1 간격으로 100 개 x 좌표 리스트 생성
y = np.sin(x)              # sin 함수, y 좌표, x와 같은 개수의 리스트 생성

plt.plot(x,y)
plt.text(6, -0.75, f'sin(x) graph', fontdict={'size': 20, 'color': 'red'})
plt.title(f'Graph')
plt.xlabel('x')
plt.ylabel('f(x)')

plt.show()
print('x의 개수:', len(x))
print('y의 개수:', len(y))
```
